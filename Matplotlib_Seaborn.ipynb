# What is Matplotlib?
Matplotlib is a low level graph plotting library in python that serves as a visualization utility.

Matplotlib was created by John D. Hunter.

Matplotlib is open source and we can use it freely.

Matplotlib is mostly written in python, a few segments are written in C, Objective-C and Javascript for Platform compatibility.

# What is Seaborn?
Seaborn is a library that uses Matplotlib underneath to plot graphs. It will be used to visualize random distributions.

```python
import matplotlib as plt

print(plt.__version__)
```

Draw a line in a diagram from position (0,0) to position (6,250):

```python
import matplotlib.pyplot as plt
import numpy as np

xpoints = np.array([0, 6])
ypoints = np.array([0, 250])

plt.plot(xpoints, ypoints)
plt.show()
```

# Plotting x and y points

The plot() function is used to draw points (markers) in a diagram.

By default, the plot() function draws a line from point to point.

The function takes parameters for specifying points in the diagram.

Parameter 1 is an array containing the points on the x-axis.

Parameter 2 is an array containing the points on the y-axis.

If we need to plot a line from (1, 3) to (8, 10), we have to pass two arrays [1, 8] and [3, 10] to the plot function.

Plotting without line

```python
import matplotlib.pyplot as plt
import numpy as np

xpoints = np.array([1, 8])
ypoints = np.array([3, 10])

plt.plot(xpoints, ypoints, 'o')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

ypoints = np.array([3, 8, 1, 10, 5, 7])


# By default, the xpoints get values like 0, 1, 2, 3, 4...
plt.plot(ypoints)
plt.show()
```

```python
import matplotlib.pyplot as plt

x=[1,2,3]
y=[80,90,85]
plt.plot(x,y)
plt.title('Line chart')
plt.xlabel('student')
plt.ylabel('marks')
plt.show()
```

```python
import matplotlib.pyplot as plt

x=["A", "B", "C"]
y=[80,90,85]
plt.bar(x,y)
plt.title('Bar chart')
plt.xlabel('student')
plt.ylabel('marks')
plt.show()
```

```python
x=[10,20,40,13,14,12,12,11,13,14,15,18,24]
plt.hist(x,bins=10,color='orange')
plt.title('historgrams')
plt.xlabel('marks')
plt.ylabel('freq')

plt.show()
```

```python
x=[10,20,40,30]
y=[80,90,85,95]
plt.scatter(x,y)
plt.title('Scatterplot')
plt.xlabel('Study hours')
plt.ylabel('Marks')
plt.legend(['avg values'])
plt.show()
```

```python
cars=['Maruti',"Suzuki",'BMW','Ford']
data=[27,33,22,18]
plt.pie(data,labels=cars,autopct='%1.1f%%')
plt.title("pie chart")
plt.show()
```

```python
data=[10,20,12,14,16,50,25,33,19,27]
plt.boxplot(data)
plt.title('box plot')
plt.show()
```

```python
import pandas as pd

students = pd.DataFrame({
    'Height': [150, 160, 155, 170, 165],
    'Weight': [45, 55, 50, 70, 65],
    'Gender': ['Female', 'Male', 'Female', 'Male', 'Male']
})
print(students)
print()

import seaborn as sns
sns.scatterplot(
    x=students['Height'],
    y=students['Weight'],
    hue=students['Gender']
).set(title='Students')
```
