Tuple:

You can create a tuple by:
- Using a tuple constructor
- By listing the items inside round brackets

```python
thistuple = ('apple', 'banana', 'cherry', True, 1)
print(type(thistuple))
print(thistuple)
```

- Tuples allow duplicate values

```python
thistuple = ('apple', 'apple', 'banana', 'cherry')
print(thistuple)
```

Tuple Methods
- len


```python
thistuple = ('apple', 'banana', 'cherry')
print(len(thistuple))
```

Accessing Tuples:
- Use normal indexing
- Use negative indexing
- Use range of indexes (like string slicing)

```python
thistuple = ("thirdlast", "secondlast", "last")

# Normal indexing
print(thistuple[0])


print()
# Use range of indexes
print(thistuple[0:3])
print(thistuple[0:2])
print(thistuple[:2]) # same as above
print(thistuple[1:])
print(thistuple[1:3]) # same as above
print(thistuple[:])  # this simply creates a new copy of the tuple

print()
# Use negative indexing
print(thistuple[-1])
print(thistuple[-2])
print(thistuple[-3])

print()
# Use range of negative indices
thistuple = (1, 2, 3, 4, 5, 6, 7)

print(thistuple[-7: 0])
print(thistuple[-7: -1])
print(thistuple[-7:])
```

How do you know if an item exists in a tuple?

Answer: Use the in membership operator

```python
thistuple = (1, 2, "apple", "mango")
if "apple" in thistuple:
  print(True)
```

Although tuples are immutable, you can still modify them by converting them to a list using a list constructor.

Without the list constructor, you can natively join(append) two tuples together with the + operator, you can also multiply the content of a tuple any number of times with the * operator.

Tuple Methods:
- count()
    - How many times has a specific value occured in the tuple?

 - index()
    - What is the index of a given specified value

```python
thistuple = (1, 1, 1, 2, 1, 4, 3, 6)

print(thistuple.count(1))
# the value 1 has 4 occurences

print(thistuple.index(1))
# the value 1 has occurences at index 0, 1, 2, 4
# thus only the first occurence index is returned
```
